# Part A — Python

## A1. Variables & types

A *variable* is a labeled box that holds a value. Python figures out the **type** for you.

````python
name = "Ada"          # str  (text)
age = 16              # int  (whole number)
height_m = 1.63       # float (decimal)
is_student = True     # bool (True/False)

print(name, age, height_m, is_student)
print(type(name), type(age), type(height_m), type(is_student))
````

````python
# 🔧 Your turn: ask for a name and age, print how old they'll be in 10 years.
# Hint: input() always gives you a str — you must int(...) it to do math.
your_name = input("Your name: ")
your_age = int(input("Your age: "))
print(your_name + ", in 10 years you'll be " + str(your_age + 10) + ".")
````

## A2. Strings & f-strings

Gluing strings with `+` and `str(...)` gets ugly fast. **f-strings** are the clean way: put an `f` before the quote and drop variables inside `{ }`.

````python
name = "Ada"
age = 16
# The same sentence, the clean way:
print(f"{name} will be {age + 10} in ten years.")

# f-strings can do math and formatting inline:
print(f"Half of {age} is {age / 2}")
print(f"Pi to 2 decimals: {3.14159:.2f}")
````

## A3. Lists & `for` loops

A *list* is an ordered collection. A `for` loop walks through it one item at a time.

````python
models = ["gpt-oss", "qwen3", "gemma"]
print("We have", len(models), "models")

for m in models:
    print("-", m)

# range(n) gives you 0,1,2,...,n-1
for i in range(5):
    print("count:", i)
````

````python
# 🔧 Your turn: print the first 20 numbers of the Fibonacci sequence.
# Each number is the sum of the previous two: 0, 1, 1, 2, 3, 5, 8, ...
a, b = 0, 1
for _ in range(20):
    print(a, end=" ")
    a, b = b, a + b
````

## A4. Dictionaries

A *dict* maps **keys** to **values** — like a contacts list (name → phone number).

````python
# model name -> a short note about it
models = {
    "gpt-oss": "120B, great default",
    "qwen3": "397B, frontier reasoning",
    "gemma": "31B, small and fast",
}

# look one up by key:
print(models["qwen3"])

# walk through all of them:
for model_name, note in models.items():
    print(f"{model_name:10} -> {note}")
````

````python
# 🔧 Your turn: add a new model to the dict, then print how many models you have.
models["qwen3-small"] = "27B, tiny and quick"
print(models)
print("Total models:", len(models))
````

## A5. Making decisions — `if` / `elif` / `else`

````python
score = 15

if score >= 18:
    print("Excellent")
elif score >= 12:
    print("Solid")
else:
    print("Keep going")

# comparisons give you booleans:
print(10 > 3, 10 == 10, "a" in "cat")
````

## A6. `while` loops

A `for` loop runs a known number of times. A `while` loop runs **until a condition turns False** — perfect for "keep going until the user quits" (sound familiar? that's your chatbot).

````python
# countdown
n = 5
while n > 0:
    print(n)
    n = n - 1
print("Liftoff!")
````

````python
# 🔧 Your turn: keep asking for a word until the user types 'stop'.
while True:
    word = input("Say something (or 'stop'): ")
    if word == "stop":
        break
    print("You said:", word)
````

## A7. Functions

A *function* is a reusable machine: give it inputs, get an output. Define once with `def`, call it as many times as you like.

````python
def greet(person):
    return f"Hello, {person}!"

print(greet("Ada"))
print(greet("the cohort"))

def add(a, b):
    return a + b

print(add(2, 3))
````

````python
# 🔧 Your turn: write is_prime(n), then print every prime under 30.
def is_prime(n):
    if n < 2:
        return False
    for d in range(2, n):
        if n % d == 0:
            return False
    return True

for n in range(30):
    if is_prime(n):
        print(n, end=" ")
````

## A8. Imports

You don't write everything yourself — Python ships with batteries. `import` pulls in extra tools.

````python
import random
import math

print("Random dice roll:", random.randint(1, 6))
print("Square root of 144:", math.sqrt(144))

# import just one thing:
from random import choice
print("Pick a model:", choice(["gpt-oss", "qwen3", "gemma"]))
````

## A9. When things break — `try` / `except`

Code fails: bad input, network down, wrong token. `try`/`except` catches the error so your program keeps its cool instead of crashing.

````python
try:
    number = int(input("Type a number: "))
    print("Half of that is", number / 2)
except ValueError:
    print("That wasn't a number! But I didn't crash. 😎")
````

This is exactly how you'll harden your LLM calls in `chat.py`:
```python
try:
    response = client.chat.completions.create(...)
except Exception as e:
    print(f"Oops, something broke: {e}")
```

## A10. Reading & writing files

Your chatbot can save its conversation. Files are how.

````python
# write a file
with open("notes.txt", "w") as f:
    f.write("first line\n")
    f.write("second line\n")

# read it back
with open("notes.txt") as f:
    contents = f.read()

print(contents)
````

## B1. The core loop (run these in your terminal)

This is the exact loop you'll use all summer:

```bash
git checkout -b week02/yourpair-chat   # 1. make a branch to work on

```

Then on GitHub you click **"Compare & pull request"** to open a **PR**.

## B2. Pull requests & review

A **Pull Request (PR)** says "please merge my branch into `main`." Your pair partner **reviews** it — reads the code, leaves comments, clicks approve. *Then* it merges. **Never push straight to `main`.**

## B3. `.gitignore`

Some files should *never* be committed — your `.env` (secret token!), junk like `__pycache__/`. List them in a file called `.gitignore` and git pretends they don't exist:

```
.env
__pycache__/
practice/
```

# Part C — Put it together: `chat.py` 🤝

Everything above combines into this week's milestone: a command-line chatbot. Read the scaffold below — you'll recognize **every** piece now (imports, `while True`, `input()`, a list of messages, try/except waiting to be added).

Build it with your pair in your own file. (Running it needs your NRP token in a `.env`, so do that in your terminal, not here.)

````python
# chat.py  —  study this; build your own version with your pair.
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
client = OpenAI(
    api_key=os.environ["NRP_LLM_TOKEN"],
    base_url=os.environ["NRP_LLM_BASE_URL"],
)

print("NRP Chat. Type 'quit' to exit.")
history = []                                   # <- the memory (A3 list, A6 while, A4 dicts inside)
while True:
    user_input = input("You: ")
    if user_input.lower() in ("quit", "exit"):
        break
    history.append({"role": "user", "content": user_input})
    response = client.chat.completions.create(model="gpt-oss", messages=history)
    reply = response.choices[0].message.content
    history.append({"role": "assistant", "content": reply})
    print("Bot:", reply, "\n")
````
